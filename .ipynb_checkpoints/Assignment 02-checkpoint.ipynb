# Q#07

```python
num1 = float(input("Enter first number: "))
operator = input("Enter operator (+, -, x, /): ")
num2 = float(input("Enter second number: "))

if operator == "+":
    result = num1 + num2

elif operator == "-":
    result = num1 - num2

elif operator == "x":
    result = num1 * num2

elif operator == "/":
    result = num1 / num2

else:
    result = "Invalid operator"

print("Result:", result)
```

# Q#09

```python
num = int(input("Enter an integer number: "))

if num % 2 == 0 and num % 3 == 0:
    print("The number is divisible by BOTH 2 and 3.")
elif num % 2 == 0:
    print("The number is divisible by 2.")
elif num % 3 == 0:
    print("The number is divisible by 3.")
else:
    print("The number is NOT divisible by 2 or 3.")
```

# Q#13

```python
salary= 3000000

if salary>2500000:
    tax_amount = salary*0.15
elif salary > 1500000 and salary < 2500000:
    tax_amount = salary*0.1
elif salary >500000 and salary <1500000:
    tax_amount = salary * 0.05
else:
    tax_amount = salary*0

gross_salary = salary- tax_amount
print(f"Tax deduction {tax_amount}")
print(f"Gross deduction {gross_salary}")
```
