# Question No 7

```python
num1 = float(input("Enter first number: "))
operator = input("Enter operator (+, -, x, /): ")
num2 = float(input("Enter second number: "))

if operator == "+":
    result = num1 + num2

elif operator == "-":
    result = num1 - num2

elif operator == "x":
    result = num1 * num2

elif operator == "/":
    result = num1 / num2

else:
    result = "Invalid operator"

print("Result:", result)
```

# Question No 9

```python
number = int(input("Enter a number: "))

if number % 2 == 0 and number % 3 == 0:
    print("Number is divisible by both 2 and 3")
elif number % 2 == 0:
    print("Number is divisible by 2")
elif number % 3 == 0:
    print("Number is divisible by 3")
else:
    print("Number is not divisible by 2 or 3")
```

# Question No 11

```python
temperature = float(input("Enter temperature in Celsius: "))

if temperature < 0:
    print("Freezing")
elif temperature > 0 and temperature < 26:
    print("Moderate")
elif temperature > 26:
    print("Hot")
else:
    print("Temperature is 0 or 26")
```

# Question No 12

```python
salary = float(input("Enter your salary: "))

if salary > 600000:
    tax = salary * 5 / 100
    gross_salary = salary - tax
    print("Tax:", tax)
    print("Gross Salary:", gross_salary)
else:
    print("No tax deduction")
    print("Gross Salary:", salary)
```

# Question No 13

```python
salary= 5000000

if salary>2500000:
    tax_amount = salary*0.15
elif salary > 1500000 and salary < 2500000:
    tax_amount = salary*0.1
elif salary >500000 and salary <1500000:
    tax_amount = salary * 0.05
else:
    tax_amount = salary*0

gross_salary = salary- tax_amount
print(f"Tax deduction {tax_amount}")
print(f"Gross deduction {gross_salary}")
```
